```python
db = client["PawConnect"]
users = db["users"]

user = {
    "name": "Madhumithaa",
    "email": "madhumithaa@example.com",
    "password": "12345"
}

result = users.insert_one(user)

print("User inserted successfully")
print("User ID:", result.inserted_id)
```

```python
users_list = [
    {
        "name": "Arun",
        "email": "arun@example.com",
        "password": "12345"
    },
    {
        "name": "Priya",
        "email": "priya@example.com",
        "password": "12345"
    },
    {
        "name": "Rahul",
        "email": "rahul@example.com",
        "password": "12345"
    }
]

result = users.insert_many(users_list)

print("Users inserted successfully")
print("Number of users inserted:", len(result.inserted_ids))
```

```python
for user in users.find():
    print(user)
```

```python
user = users.find_one({"email": "madhumithaa@example.com"})

print("Particular User:")
print(user)
```

```python
result = users.update_one(
    {"email": "madhumithaa@example.com"},
    {"$set": {"name": "Madhumithaa P"}}
)

print("Record updated successfully")
print("Number of records updated:", result.modified_count)
```

```python
result = users.delete_one(
    {"email": "madhumithaa@example.com"}
)

print("Record deleted successfully")
print("Number of records deleted:", result.deleted_count)
```
